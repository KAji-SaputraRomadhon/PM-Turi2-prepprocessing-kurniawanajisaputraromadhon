## Dataset

```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
print("Nilai hilang sebelum imputasi:")
print(df.isna().sum())
df
```

## Langkah 1

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())

display(df)
```

## Langkah 2

```python
X = df.drop(columns=['membeli'])
y = df['membeli']

display(X)
display(y)
```

## Langkah 3

```python
# Ordinal encoding pendidikan
X['pendidikan'] = X['pendidikan'].map({
    'SMA': 0,
    'S1': 1,
    'S2': 2
})

# One-hot encoding kota
X = pd.get_dummies(
    X,
    columns=['kota'],
    dtype=int
)

# Encoding target
y = y.map({
    'Tidak': 0,
    'Ya': 1
})

display(X)
display(y)
```

## Langkah 4

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

display(X_train)
display(X_test)
```

## Langkah 5

```python
# Langkah 5 — Penskalaan fitur

from sklearn.preprocessing import StandardScaler

num = ['usia', 'pendapatan', 'pendidikan']

sc = StandardScaler()

X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])

display(X_train.round(4))
display(X_test.round(4))
```

## Langkah 6

```python
# Langkah 6 — Verifikasi

print("Missing values:")
print(X_train.isnull().sum())
print(X_test.isnull().sum())

print("\nData types:")
print(X_train.dtypes)

print("\nStatistik data training:")
display(X_train.describe().round(4))
```

```python
from sklearn.preprocessing import MinMaxScaler

X_minmax = X.copy()

X_train_mm, X_test_mm, y_train_mm, y_test_mm = train_test_split(
    X_minmax,
    y,
    test_size=0.3,
    random_state=42
)

num = ['usia', 'pendapatan', 'pendidikan']

mm = MinMaxScaler()


X_train_mm[num] = mm.fit_transform(X_train_mm[num])
X_test_mm[num] = mm.transform(X_test_mm[num])


print("Statistik X_train setelah MinMaxScaler:")
display(X_train_mm.describe().round(4))

print("Statistik X_test setelah MinMaxScaler:")
display(X_test_mm.describe().round(4))
display(X_train_mm.round(4))
```

4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.

fit_transform() digunakan pada data training karena scaler perlu mempelajari parameter dari data training, seperti nilai rata-rata dan standar deviasi pada StandardScaler.

## Refleksi

1. Mengapa urutan 'split dulu, baru scaling' penting?

Untuk mencegah terjadinya data leakage. Data dibagi menjadi data latih dan data uji terlebih dahulu. Setelah itu, StandardScaler melakukan fit_transform() hanya pada data latih untuk mempelajari parameter scaling. Data uji kemudian hanya menggunakan transform() dengan parameter yang sudah diperoleh dari data latih.

2. Kapan label encoding dan kapan one-hot encoding?

Label encoding digunakan ketika kategori memiliki urutan atau tingkatan (ordinal).
Sedangkan one-hot encoding digunakan untuk kategori nominal, yaitu kategori yang tidak memiliki tingkatan.

3. Apa perbedaan normalisasi dan standardisasi?



* Normalisasi merupakan proses mengubah nilai data ke rentang tertentu, misalnya 0 sampai 1.
* Standardisasi merupakan proses mengubah data berdasarkan rata-rata dan standar deviasi.
